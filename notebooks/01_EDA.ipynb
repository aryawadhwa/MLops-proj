# Exploratory Data Analysis: Meter BR28

This notebook performs EDA on the processed daily electricity consumption dataset for meter BR28.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
```

```python
# Load the dataset
df = pd.read_csv('../data/processed/bareilly_br28_daily.csv')
df['date'] = pd.to_datetime(df['date'])
df = df.set_index('date')

df.head()
```

## 1. Time Series Visualization

```python
plt.figure(figsize=(15, 5))
plt.plot(df.index, df['daily_kwh'], linewidth=1)
plt.title('Daily Electricity Consumption for BR28')
plt.xlabel('Date')
plt.ylabel('Daily Consumption (kWh)')
plt.tight_layout()
plt.show()
```

## 2. Missing Dates Analysis

```python
expected_dates = pd.date_range(start=df.index.min(), end=df.index.max(), freq='D')
missing_dates = expected_dates.difference(df.index)

print(f"Total expected days: {len(expected_dates)}")
print(f"Total observed days: {len(df)}")
print(f"Missing days: {len(missing_dates)}")

if len(missing_dates) > 0:
    print("\nFirst few missing dates:")
    print(missing_dates[:5])
```

## 3. Distribution of Daily Consumption

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['daily_kwh'], bins=50, kde=True)
plt.title('Distribution of Daily Consumption')
plt.xlabel('Daily Consumption (kWh)')
plt.show()
```
